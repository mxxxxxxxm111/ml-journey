```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Построй на одном графике sin(x) и cos(x) для x = np.linspace(0, 10, 100): заголовок, подписи осей, легенда, сетка.
x = np.linspace(0,10, 100)
fig, ax = plt.subplots(figsize = (8,4))
ax.plot(x, np.sin(x), label = 'sin')
ax.plot(x, np.cos(x), label = 'cos')
ax.set_title('Заголовок')
ax.set_xlabel('Подпись x')
ax.set_ylabel('Подпись y')

ax.legend()
ax.grid(True)
plt.show()
```

```python
#Сгенерируй rng = np.random.default_rng(42), x = rng.normal(0, 1, 100), y = 2 * x + rng.normal(0, 0.5, 100). 
# Построй диаграмму рассеяния. Облако точек должно идти по диагонали вверх
rng = np.random.default_rng(42)
x = rng.normal(0, 1 ,100)
y = 2 * x + rng.normal(0, 0.5, 100)
fig, ax = plt.subplots()
ax.scatter(x, y)
ax.set_title('Заголовок')
ax.set_xlabel('Подпись X')
ax.set_ylabel('Подпись Y')
plt.show()
```

```python
#Гистограмма 1000 чисел из нормального распределения (bins=30) и вертикальная линия на их среднем через axvline.
data = rng.normal(0, 1, 1000)
fig, ax = plt.subplots()
ax.axvline(data.mean(),color="red", label="среднее")
ax.hist(data,bins = 30)
ax.legend()
plt.show()
```

```python
#Столбчатая диаграмма по словарю {"еда": 650, "транспорт": 60, "одежда": 300, "кино": 150}: столбцы по убыванию, подписи осей и заголовок.
expenses = {"еда": 650, "транспорт": 60, "одежда": 300, "кино": 150}
fig, ax = plt.subplots()
s = pd.Series(expenses).sort_values(ascending=False)
ax.bar(s.index, s.values)
ax.set_title('Заголовок')
ax.set_ylabel('категория')
ax.set_xlabel('сумма, руб')
plt.show()
```

```python
#Фигура 1×2: слева гистограмма из задачи 3, справа диаграмма из задачи 2. Сохрани её в plots.png.
rng = np.random.default_rng(42)
fig, axes = plt.subplots(1,2,figsize=(10, 4))
data1 = rng.normal(0, 1, 1000)
x1 = rng.normal(0, 1 ,100)
y1 = 2 * x1 + rng.normal(0, 0.5, 100)
axes[0].hist(data1,bins = 30)
axes[1].scatter(x1, y1)
axes[0].set_title('Гистограмма')
axes[1].set_title('Диаграмма')
fig.tight_layout()
fig.savefig("plots.png", dpi=150)
plt.show()
```

```python
#Создай pd.DataFrame из словаря: имена пяти студентов, их группы ("A" или "B") и оценки. 
import pandas as pd
# Выведи студентов с оценкой выше 70, среднюю оценку по группам и таблицу, отсортированную по убыванию оценки.
d = pd.DataFrame({'student_name':['Ivan','Vasyan', 'Oleg','Igor','Lexa'],
                  'Group':['A','B','B','A','A'],
                  'marks':[50,20,90,65,71]})
print(d[d['marks'] > 70]['student_name'])
print(d.groupby('Group')['marks'].mean())
print(d.sort_values('marks', ascending=False))


```

```python
#На том же DataFrame: сколько студентов в каждой группе (value_counts) и максимум и минимум оценки по группам (groupby с agg(["max", "min"])).
d = pd.DataFrame({'student_name':['Ivan','Vasyan', 'Oleg','Igor','Lexa'],
                  'Group':['A','B','B','A','A'],
                  'marks':[50,20,90,65,71]})
print(d['Group'].value_counts())
print(d.groupby('Group')['marks'].agg(['max', 'min']))
```
